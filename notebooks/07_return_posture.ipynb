# ۷. حالت بدن در ویزیت بازگشت سمعی

این قسمت ۳ است. فقط SBP سمعی. مدل یک بار روی ویزیت بازگشت آموزش می‌بیند و بعد خطای همان پیش‌بینی به سه بخش تقسیم می‌شود: همه، خوابیده، نشسته. مقاله در جدول VII فقط انحراف معیار را داده و ستون «همه» با جدول V یکی است.

نام‌ها خودشان کلمهٔ supine ندارند. قاعده این است: اگر نام نشسته باشد، نشسته است؛ اگر Supine داشته باشد، خوابیده است؛ وگرنه «challenge start» یا «calibration start» خوابیده حساب می‌شود. بقیه در ستون همه می‌مانند و در دو ستون حالت نمی‌آیند. این قاعده تصمیم ماست، نه نص مقاله.

جدول VII برای کل جمعیت، انحراف معیار تونومتری را ۹٫۷۵، ۹٫۲۷ و ۱۰٫۵ برای همه، خوابیده و نشسته می‌دهد. انتظار این است که خوابیده از نشسته کوچک‌تر باشد.

```python
import sys
from pathlib import Path
sys.path.insert(0, '/content/aurorabp_reproduction')
import pandas as pd
from src.bp_protocol import run_return_posture

candidates = [
    Path('/content/aurora_waveforms'),
    Path('/content/drive/MyDrive/aurora_dataset'),
]
data_dir = next(path for path in candidates if (path / 'features.tsv').is_file() and (path / 'participants.tsv').is_file())
print('داده:', data_dir)
features = pd.read_csv(data_dir / 'features.tsv', sep='\t')
participants = pd.read_csv(data_dir / 'participants.tsv', sep='\t')
output = Path('/content/aurorabp_outputs/return_posture.json')
output.parent.mkdir(parents=True, exist_ok=True)
results = run_return_posture(features, participants, output_path=str(output))
print(output)
results
```

```python
show = results.loc[:, [
    'population', 'row_set', 'model', 'posture', 'n_rows', 'n_participants', 'boot_bias', 'boot_sd', 'reliability',
]].copy()
for column in ('boot_bias', 'boot_sd'):
    show[column] = show[column].map(lambda value: round(float(value), 2) if pd.notna(value) else value)
print('ستون مقایسه با جدول VII، boot_sd است.')
show
```
