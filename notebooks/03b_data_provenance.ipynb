# ۳ب. قیف data provenance قبل از مدل

این نوت‌بوک را قبل از دفترچه‌های ۰۴ تا ۰۷ اجرا کنید. مدل فقط جعبهٔ analyzed را می‌بیند.

شکل `data_provenance.png` در مخزن رسمی این جعبه‌ها را دارد: سمعی ۶۷۲ هم‌تراز و ۶۴۲ تحلیل‌شده، اسیلومتریک ۵۴۸ هم‌تراز و ۴۸۳ تحلیل‌شده.

رضایت، انصراف، خرابی دستگاه، خرابی فایل، و شکست هم‌ترازی دو ECG در فایل منتشرشده نیستند. پوشه‌های موج از جعبهٔ هم‌تراز شروع می‌شوند. در موجودی قبلی، پوشهٔ موج سمعی ۶۷۲ و اسیلومتریک ۵۴۸ بود.

جعبهٔ analyzed همان `in_feature_table` است. کسی که موج دارد و در این فهرست نیست، حذف کیفیت همان شکل است. برچسب «۳۱» روی شکل با تفاضل ۶۷۲ و ۶۴۲ یکی نیست؛ تفاضل ۳۰ است. این نوت‌بوک عدد مشاهده‌شده را چاپ می‌کند و برای رسیدن به برچسب، کسی را حذف نمی‌کند.

حداقل ۱۷ اندازه‌گیری شکل ۷ بعد از این قیف است. این نوت‌بوک آن تعداد را ۲۲۷ نمی‌کند.

```python
import json
import sys
from pathlib import Path
sys.path.insert(0, '/content/aurorabp_reproduction')
import pandas as pd
from src.bp_protocol import provenance_report, restrict_to_analyzed

data_dir = Path('/content/drive/MyDrive/aurora_dataset')
participants = pd.read_csv(data_dir / 'participants.tsv', sep='\t')
features = pd.read_csv(data_dir / 'features.tsv', sep='\t')
auscultatory = pd.read_csv(data_dir / 'measurements_auscultatory.tsv', sep='\t')
oscillometric = pd.read_csv(data_dir / 'measurements_oscillometric.tsv', sep='\t')
report = provenance_report(participants, features, auscultatory, oscillometric)
kept = restrict_to_analyzed(features, participants)
output = Path('/content/aurorabp_outputs')
output.mkdir(parents=True, exist_ok=True)
pd.DataFrame({'pid': report['analyzed_ids']}).to_csv(output / 'analyzed_pids.tsv', sep='\t', index=False)
summary = {key: value for key, value in report.items() if key != 'analyzed_ids'}
(output / 'data_provenance.json').write_text(json.dumps(summary, indent=2), encoding='utf-8')
print(output / 'analyzed_pids.tsv')
print('feature rows after the analyzed gate:', len(kept))
```
