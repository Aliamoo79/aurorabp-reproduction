```python
import matplotlib.pyplot as plt
from src.filters import stage_response

panels = [
    ('ECG', [('ecg', 'dc'), ('ecg', 'lowpass'), ('ecg', 'notch')], [0.1, 40, 45, 60], 80),
    ('PPG', [('ppg', 'highpass'), ('ppg', 'lowpass')], [0.25, 10, 12], 30),
    ('Tonometry', [('tonometry', 'highpass'), ('tonometry', 'lowpass')], [0.2, 0.3, 22, 26], 40),
]
figure, axes = plt.subplots(3, 1, figsize=(8, 9))
for axis, (title, stages, marks, limit) in zip(axes, panels):
    for kind, stage in stages:
        response = stage_response(kind, stage)
        axis.plot(response['hz'], response['db'], label=stage)
    for mark in marks:
        axis.axvline(mark, color='0.6', lw=0.6, ls='--')
    axis.set_xlim(0, limit)
    axis.set_ylim(-80, 5)
    axis.set_title(title)
    axis.set_ylabel('Gain (dB)')
    axis.grid(True, alpha=0.3)
    axis.legend(loc='lower left', fontsize=8)
axes[-1].set_xlabel('Frequency (Hz)')
figure.tight_layout()
destination = PROJECT / 'outputs' / 'figures' / 'filter_responses.png'
destination.parent.mkdir(parents=True, exist_ok=True)
figure.savefig(destination, dpi=140)
figure
```
