```python
import sys
from pathlib import Path
sys.path.insert(0, '/content/aurorabp_reproduction')
import pandas as pd
from src.bp_protocol import resting_reference_sbp, subset_ids, run_ambulatory_subsets

candidates = [
    Path('/content/aurora_waveforms'),
    Path('/content/drive/MyDrive/aurora_dataset'),
]
data_dir = next(path for path in candidates if (path / 'features.tsv').is_file() and (path / 'participants.tsv').is_file())
print('داده:', data_dir)
features = pd.read_csv(data_dir / 'features.tsv', sep='\t')
participants = pd.read_csv(data_dir / 'participants.tsv', sep='\t')
resting = resting_reference_sbp(features)
print('شرکت‌کننده با SBP استراحت:', int(resting.size))
print('نرمال با آستانه ۱۳۰:', int((resting <= 130).sum()))
print('سن ۳۵ یا کمتر:', int(subset_ids(features, participants, 'young').size))
output = Path('/content/aurorabp_outputs/ambulatory_24h_subsets.json')
output.parent.mkdir(parents=True, exist_ok=True)
results = run_ambulatory_subsets(features, participants, output_path=str(output))
print(output)
results
```

```python
show = results.loc[:, [
    'population', 'row_set', 'model', 'target', 'n_rows', 'n_participants',
    'boot_bias', 'boot_sd', 'lopo_bias', 'lopo_sd', 'reliability',
]].copy()
for column in ('boot_bias', 'boot_sd', 'lopo_bias', 'lopo_sd'):
    show[column] = show[column].map(lambda value: round(float(value), 2) if pd.notna(value) else value)
print('boot_bias و boot_sd عدد گزارش‌اند.')
show
```
